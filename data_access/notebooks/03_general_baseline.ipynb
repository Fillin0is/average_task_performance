```python
import os
import warnings
from datetime import UTC, datetime
from time import perf_counter

os.environ.setdefault("LOKY_MAX_CPU_COUNT", "1")
warnings.filterwarnings(
    "ignore",
    message="Could not find the number of physical cores.*",
    category=UserWarning,
)

import joblib
import numpy as np
import pandas as pd
from sqlalchemy import text

from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import (
    mean_absolute_error,
    median_absolute_error,
    root_mean_squared_log_error,
    root_mean_squared_error,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder

from data_access.connection import engine
from data_access.config import project_path


query_path = project_path.query("training_dataset.sql")
```

```python
query = text(query_path.read_text(encoding="utf-8"))

issue_df = pd.read_sql_query(query, engine)

print("Количество строк:", len(issue_df))
issue_df.head()
```

## Хронологическое разделение

Модель обучается на первых 70% задач. Следующие 15% используются как validation, последние 15% — как test.

```python
issue_df = issue_df.sort_values(
    ["Creation_Date", "ID"]
).reset_index(drop=True)

train_end = int(len(issue_df) * 0.70)
validation_end = int(len(issue_df) * 0.85)

train_df = issue_df.iloc[:train_end].copy()
validation_df = issue_df.iloc[train_end:validation_end].copy()
test_df = issue_df.iloc[validation_end:].copy()

pd.DataFrame(
    {
        "rows": [len(train_df), len(validation_df), len(test_df)],
        "date_min": [
            train_df["Creation_Date"].min(),
            validation_df["Creation_Date"].min(),
            test_df["Creation_Date"].min(),
        ],
        "date_max": [
            train_df["Creation_Date"].max(),
            validation_df["Creation_Date"].max(),
            test_df["Creation_Date"].max(),
        ],
    },
    index=["train", "validation", "test"],
)
```

## Формирование `X` и `y`

```python
feature_columns = [
    "Description_Issue_Length",
    "Description_Project_Length",
    "Comment_Count",
    "Comment_Total_Length",
    "Change_Count",
    "Initial_Type",
    "Initial_Priority",
]
target_column = "Duration_Days"

X_train = train_df[feature_columns].copy()
y_train = train_df[target_column].copy()

X_validation = validation_df[feature_columns].copy()
y_validation = validation_df[target_column].copy()

X_test = test_df[feature_columns].copy()
y_test = test_df[target_column].copy()

pd.DataFrame(
    {
        "X_shape": [X_train.shape, X_validation.shape, X_test.shape],
        "y_shape": [y_train.shape, y_validation.shape, y_test.shape],
    },
    index=["train", "validation", "test"],
)
```

## Preprocessing

`Description_Issue_Length` и `Description_Project_Length` передаются как числа. `OrdinalEncoder` превращает строковые тип и приоритет в числовые категории. Неизвестная категория при запросе получает `-1`.

```python
preprocessing = ColumnTransformer(
    transformers=[
        (
            "numeric",
            "passthrough",
            [
                "Description_Issue_Length",
                "Description_Project_Length",
                "Comment_Count",
                "Comment_Total_Length",
                "Change_Count",
            ],
        ),
        (
            "categories",
            OrdinalEncoder(
                handle_unknown="use_encoded_value",
                unknown_value=-1,
                encoded_missing_value=-1,
            ),
            ["Initial_Priority", "Initial_Type"]
        ),
    ],
    verbose_feature_names_out=False,
)

preprocessing
```

```python
regressor = HistGradientBoostingRegressor(
    categorical_features=[
        False, False, False, False, False,
        True, True,
    ],
    learning_rate=0.08,
    max_iter=250,
    max_leaf_nodes=31,
    min_samples_leaf=50,
    l2_regularization=1.0,
    random_state=42,
)

pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessing),
        ("regressor", regressor),
    ]
)

model = TransformedTargetRegressor(
    regressor=pipeline,
    func=np.log1p,
    inverse_func=np.expm1,
    check_inverse=False,
)

model
```

```python
started_at = perf_counter()
model.fit(X_train, y_train)
training_seconds = perf_counter() - started_at

print(f"Обучение завершено за {training_seconds:.2f} сек.")
```

```python
def calculate_metrics(y_true, prediction):
    prediction = np.maximum(prediction, 0.0)
    return {
        "MAE_days": mean_absolute_error(y_true, prediction),
        "Median_AE_days": median_absolute_error(y_true, prediction),
        "RMSE_days": root_mean_squared_error(y_true, prediction),
        "RMSLE": root_mean_squared_log_error(y_true, prediction),
    }
```

```python
metrics = {}

for split_name, features, target in (
    ("train", X_train, y_train),
    ("validation", X_validation, y_validation),
    ("test", X_test, y_test),
):
    prediction = model.predict(features)
    metrics[split_name] = calculate_metrics(target, prediction)
```

## Медианный baseline на тестовой выборке

```python
median_target = y_train.median()
baseline_prediction = np.full(
    len(y_test),
    median_target,
)

baseline_metrics = calculate_metrics(
    y_test,
    baseline_prediction,
)
```

## Модель без логарифмического преобразования целевой переменной

```python
from sklearn.base import clone

direct_model = clone(pipeline)
direct_model.fit(X_train, y_train)

direct_metrics = {}

for split_name, features, target in (
    ("train", X_train, y_train),
    ("validation", X_validation, y_validation),
    ("test", X_test, y_test),
):
    prediction = direct_model.predict(features)
    direct_metrics[split_name] = calculate_metrics(
        target,
        prediction,
    )
```

## Сравнение моделей с медианным baseline на test

```python
test_comparison = pd.DataFrame(
    {
        "log_target": metrics["test"],
        "direct": direct_metrics["test"],
        "median_baseline": baseline_metrics,
    }
).T

test_comparison["MAE_Improvement"] = (
    1
    - test_comparison["MAE_days"]
    / baseline_metrics["MAE_days"]
)

test_comparison["RMSE_Improvement"] = (
    1
    - test_comparison["RMSE_days"]
    / baseline_metrics["RMSE_days"]
)

display(
    test_comparison.style.format({
        "MAE_days": "{:.4f}",
        "Median_AE_days": "{:.4f}",
        "RMSE_days": "{:.4f}",
        "RMSLE": "{:.4f}",
        "MAE_Improvement": "{:+.2%}",
        "RMSE_Improvement": "{:+.2%}",
    })
)
```

## Сохранение модели для Backend

```python
model_path = project_path.artifact("duration_model.joblib")
model_path.parent.mkdir(parents=True, exist_ok=True)

model_bundle = {
    "model": model,
    "feature_columns": feature_columns,
    "target_column": target_column,
    "trained_at": datetime.now(UTC).isoformat(),
    "training_rows": len(X_train),
    "training_seconds": training_seconds,
    "metrics": metrics,
}

joblib.dump(model_bundle, model_path)
print("Модель сохранена:", model_path)
```
