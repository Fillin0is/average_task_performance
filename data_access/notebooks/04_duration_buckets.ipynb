```python
QUERY_PATH = PROJECT_ROOT / "data_access" / "queries" / "training_dataset.sql"

query = text(QUERY_PATH.read_text(encoding="utf-8"))   

issue_df = pd.read_sql_query(query, engine)
print("Количество строк: ", len(issue_df))

issue_df.head(10)
```

```python
issue_df = issue_df.sort_values(
    ["Creation_Date", "ID"]
).reset_index(drop=True)

train_end = int(len(issue_df) * 0.7)
validation_end = int(len(issue_df) * 0.85)

train_df = issue_df.iloc[:train_end].copy()
validation_df = issue_df.iloc[train_end:validation_end].copy()
test_df = issue_df.iloc[validation_end:].copy()

pd.DataFrame(
    {
        "count_rows": [len(train_df), len(validation_df), len(test_df)],
        "date_min": [
            train_df["Creation_Date"].min(),
            validation_df["Creation_Date"].min(),
            test_df["Creation_Date"].min()
        ],
        "date_max": [
            train_df["Creation_Date"].max(),
            validation_df["Creation_Date"].max(),
            test_df["Creation_Date"].max()
        ],
    },
    index=["train", "validation", "test"]
)
```

### Разбиение выборок по временным бакетам

```python
train_0_1 = train_df[train_df["Duration_Days"].between(0, 1)]
train_1_7 = train_df[train_df["Duration_Days"].between(1, 7, 'right')]
train_7_30 = train_df[train_df["Duration_Days"].between(7, 30, 'right')]
train_30_365 = train_df[train_df["Duration_Days"].between(30, 365, 'right')]


validation_0_1 = validation_df[validation_df["Duration_Days"].between(0, 1)]
validation_1_7 = validation_df[validation_df["Duration_Days"].between(1, 7, 'right')]
validation_7_30 = validation_df[validation_df["Duration_Days"].between(7, 30, 'right')]
validation_30_365 = validation_df[validation_df["Duration_Days"].between(30, 365, 'right')]


test_0_1 = test_df[test_df["Duration_Days"].between(0, 1)]
test_1_7 = test_df[test_df["Duration_Days"].between(1, 7, 'right')]
test_7_30 = test_df[test_df["Duration_Days"].between(7, 30, 'right')]
test_30_365 = test_df[test_df["Duration_Days"].between(30, 365, 'right')]


print("TRAIN")
print(f"  0–1 день:     {len(train_0_1)}")
print(f"  1–7 дней:     {len(train_1_7)}")
print(f"  7–30 дней:    {len(train_7_30)}")
print(f"  30–365 дней:  {len(train_30_365)}")
print(
    "  Всего по бакетам:",
    f"{sum(map(len, [train_0_1, train_1_7, train_7_30, train_30_365]))}",
    f"из {len(train_df)}",
)

print("\nVALIDATION")
print(f"  0–1 день:     {len(validation_0_1)}")
print(f"  1–7 дней:     {len(validation_1_7)}")
print(f"  7–30 дней:    {len(validation_7_30)}")
print(f"  30–365 дней:  {len(validation_30_365)}")
print(
    "  Всего по бакетам:",
    f"{sum(map(len, [validation_0_1, validation_1_7, validation_7_30, validation_30_365]))}",
    f"из {len(validation_df)}",
)

print("\nTEST")
print(f"  0–1 день:     {len(test_0_1)}")
print(f"  1–7 дней:     {len(test_1_7)}")
print(f"  7–30 дней:    {len(test_7_30)}")
print(f"  30–365 дней:  {len(test_30_365)}")
print(
    "  Всего по бакетам:",
    f"{sum(map(len, [test_0_1, test_1_7, test_7_30, test_30_365]))}",
    f"из {len(test_df)}",
)
```

### Формирование `X`, `Y` для всех временных подвыборок

```python
feature_columns = [
    "Description_Issue_Length",
    "Description_Project_Length",
    "Initial_Type",
    "Initial_Priority"
]
target_column = "Duration_Days"


X_train_0_1 = train_0_1[feature_columns].copy()
y_train_0_1 = train_0_1[target_column].copy()

X_train_1_7 = train_1_7[feature_columns].copy()
y_train_1_7 = train_1_7[target_column].copy()

X_train_7_30 = train_7_30[feature_columns].copy()
y_train_7_30 = train_7_30[target_column].copy()

X_train_30_365 = train_30_365[feature_columns].copy()
y_train_30_365 = train_30_365[target_column].copy()


X_validation_0_1 = validation_0_1[feature_columns].copy()
y_validation_0_1 = validation_0_1[target_column].copy()

X_validation_1_7 = validation_1_7[feature_columns].copy()
y_validation_1_7 = validation_1_7[target_column].copy()

X_validation_7_30 = validation_7_30[feature_columns].copy()
y_validation_7_30 = validation_7_30[target_column].copy()

X_validation_30_365 = validation_30_365[feature_columns].copy()
y_validation_30_365 = validation_30_365[target_column].copy()


X_test_0_1 = test_0_1[feature_columns].copy()
y_test_0_1 = test_0_1[target_column].copy()

X_test_1_7 = test_1_7[feature_columns].copy()
y_test_1_7 = test_1_7[target_column].copy()

X_test_7_30 = test_7_30[feature_columns].copy()
y_test_7_30 = test_7_30[target_column].copy()

X_test_30_365 = test_30_365[feature_columns].copy()
y_test_30_365 = test_30_365[target_column].copy()


pd.DataFrame(
    {
        "train": [
            X_train_0_1.shape,
            X_train_1_7.shape,
            X_train_7_30.shape,
            X_train_30_365.shape,
        ],
        "validation": [
            X_validation_0_1.shape,
            X_validation_1_7.shape,
            X_validation_7_30.shape,
            X_validation_30_365.shape,
        ],
        "test": [
            X_test_0_1.shape,
            X_test_1_7.shape,
            X_test_7_30.shape,
            X_test_30_365.shape,
        ],
    },
    index=["X_0_1", "X_1_7", "X_7_30", "X_30_365"],
)
```

## Preprocessing

`Description_Length` передаётся как число. `OrdinalEncoder` превращает строковые тип и приоритет в числовые категории. Неизвестная категория при запросе получает `-1`.

```python
preprocessing = ColumnTransformer(
    transformers=[
        (
            "numeric",
            "passthrough",
            ["Description_Issue_Length", "Description_Project_Length"],
        ),
        (
            "categories",
            OrdinalEncoder(
                handle_unknown="use_encoded_value",
                unknown_value=-1,
                encoded_missing_value=-1,
            ),
            ["Initial_Priority", "Initial_Type"]
        ),
    ],
    verbose_feature_names_out=False,
)

preprocessing
```

```python
regressor = HistGradientBoostingRegressor(
    categorical_features=[False, False, True, True],
    learning_rate=0.08,
    max_iter=250,
    max_leaf_nodes=31,
    min_samples_leaf=50,
    l2_regularization=1.0,
    random_state=42
)

pipeline = Pipeline(
    steps=[
        ("preprocessing", preprocessing),
        ("regressor", regressor),
    ]
)

model = TransformedTargetRegressor(
    regressor=pipeline,
    func=np.log1p,
    inverse_func=np.expm1,
    check_inverse=False,
)

model
```

### Обучение и метрики модели 0_1, медианное предсказение, модель без log

```python
from sklearn.base import clone

started_at = perf_counter()
model_0_1 = clone(model)
model_0_1.fit(X_train_0_1, y_train_0_1)
training_seconds = perf_counter() - started_at
print(f"Обучение завершено за {training_seconds:.2f} сек.")


def calculate_metrics(y_true, prediction):
    prediction = np.maximum(prediction, 0.0)
    return {
        "MAE_days": mean_absolute_error(y_true, prediction),
        "Median_AE_days": median_absolute_error(y_true, prediction),
        "RMSE_days": root_mean_squared_error(y_true, prediction),
        "R2": r2_score(y_true, prediction),
    }

metrics_0_1 = {}

for split_name, features, target in (
    ("train_0_1", X_train_0_1, y_train_0_1),
    ("validation_0_1", X_validation_0_1, y_validation_0_1),
):
    prediction = model_0_1.predict(features)
    metrics_0_1[split_name] = calculate_metrics(target, prediction)

pd.DataFrame(metrics_0_1).T
```

```python
median_0_1 = y_train_0_1.median()
baseline_prediction_0_1 = np.full(
    len(y_validation_0_1),
    median_0_1,
)

baseline_metrics_0_1 = calculate_metrics(
    y_validation_0_1,
    baseline_prediction_0_1,
)

baseline_metrics_0_1
```

```python
model_mae = metrics_0_1["validation_0_1"]["MAE_days"]
baseline_mae = baseline_metrics_0_1["MAE_days"]

mae_improvement = 1 - model_mae / baseline_mae

print(f"Улучшение относительно медианы: {mae_improvement:.2%}")
```

```python
direct_model_0_1 = clone(pipeline)
direct_model_0_1.fit(
    X_train_0_1,
    y_train_0_1,
)

direct_metrics_0_1 = {}

for split_name, features, target in (
    ("train_0_1", X_train_0_1, y_train_0_1),
    ("validation_0_1", X_validation_0_1, y_validation_0_1),
):
    prediction = direct_model_0_1.predict(features)
    direct_metrics_0_1[split_name] = calculate_metrics(
        target,
        prediction,
    )

pd.DataFrame(direct_metrics_0_1).T
```
