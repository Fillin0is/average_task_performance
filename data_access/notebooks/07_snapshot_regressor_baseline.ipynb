```python
from data_access.snapshot_dataset import CATEGORICAL_FEATURES


for X in (X_train_0_1, X_validation_0_1, X_test_0_1):
    for column in CATEGORICAL_FEATURES:
        X[column] = X[column].fillna("Unknown").astype(str)
```

```python
import numpy as np


from sklearn.metrics import (
    mean_absolute_error,
    median_absolute_error,
    root_mean_squared_error,
    root_mean_squared_log_error
)


def calculate_metrics(y_true, prediction):
    prediction = np.maximum(prediction, 0.0)
    return {
        "MAE_hours": mean_absolute_error(y_true, prediction),
        "Median_AE_hours": median_absolute_error(y_true, prediction),
        "RMSE_hours": root_mean_squared_error(y_true, prediction),
        "RMSLE": root_mean_squared_log_error(y_true, prediction)
    }
```

```python
from catboost import CatBoostRegressor


regressor_params = {
    "iterations": 1000,
    "depth": 6,
    "learning_rate": 0.05,
    "loss_function": "RMSE",
    "cat_features": CATEGORICAL_FEATURES,
    "random_seed": 42,
    "allow_writing_files": False,
}

regressor_24h_0_1 = CatBoostRegressor(**regressor_params)
regressor_24h_1_7 = CatBoostRegressor(**regressor_params)
regressor_24h_7_30 = CatBoostRegressor(**regressor_params)
regressor_24h_30_365 = CatBoostRegressor(**regressor_params)
```

## Regressor 24h - 0-1d

```python
regressor_24h_0_1.fit(
    X_train_0_1,
    y_train_0_1,
    eval_set=(X_validation_0_1, y_validation_0_1),
    early_stopping_rounds=100,
    use_best_model=True,
    verbose=100,
)

metrics_0_1 = {}

for split_name, features, target in (
    ("train_0_1", X_train_0_1, y_train_0_1),
    ("validation_0_1", X_validation_0_1, y_validation_0_1),
    ("test_0_1", X_test_0_1, y_test_0_1),
):
    prediction = regressor_24h_0_1.predict(features)
    metrics_0_1[split_name] = calculate_metrics(target, prediction)
```

```python
median_0_1 = y_train_0_1.median()
baseline_prediction_0_1 = np.full(
    len(y_test_0_1),
    median_0_1,
)

baseline_metrics_0_1 = calculate_metrics(
    y_test_0_1,
    baseline_prediction_0_1,
)
```

```python
direct_metrics = pd.concat(
    {
        "0-1d": pd.DataFrame(metrics_0_1).T,
        "1-7d": pd.DataFrame(metrics_1_7).T,
        "7-30d": pd.DataFrame(metrics_7_30).T,
        "30-365d": pd.DataFrame(metrics_30_365).T,
    },
    names=["Bucket", "Split"],
).round(3)

median_metrics = pd.concat(
    {
        "0-1d": pd.DataFrame(baseline_metrics_0_1).T,
        "1-7d": pd.DataFrame(baseline_metrics_1_7).T,
        "7-30d": pd.DataFrame(baseline_metrics_7_30).T,
        "30-365d": pd.DataFrame(baseline_metrics_30_365).T,
    },
    names=["Bucket", "Split"],
).round(3)


direct_metrics
```

```python
median_metrics = pd.DataFrame.from_dict(
    {
        "0-1d": baseline_metrics_0_1,
        "1-7d": baseline_metrics_1_7,
        "7-30d": baseline_metrics_7_30,
        "30-365d": baseline_metrics_30_365,
    },
    orient="index",
)

median_metrics.index.name = "Bucket"

direct_test = direct_metrics[
    direct_metrics.index.get_level_values("Split").str.startswith("test_")
].droplevel("Split")

comparison = pd.concat(
    {"CatBoost": direct_test, "Median baseline": median_metrics},
    axis=1,
)

comparison.round(3)
```

```python
import joblib

classifier_24h = joblib.load(
    project_path.artifact("classifier_24h.joblib")
)

regressors = {
    "0-1d": regressor_24h_0_1,
    "1-7d": regressor_24h_1_7,
    "7-30d": regressor_24h_7_30,
    "30-365d": regressor_24h_30_365,
}

X_pipeline = validation_24h_df[FEATURE_COLUMNS].copy()
y_pipeline = validation_24h_df[TARGET_COLUMN]

for column in CATEGORICAL_FEATURES:
    X_pipeline[column] = X_pipeline[column].fillna("Unknown").astype(str)

# Сначала выбираем класс каждой задачи.
predicted_classes = classifier_24h.predict(X_pipeline).ravel()

# Затем передаём каждую группу соответствующему регрессору.
pipeline_predictions = np.zeros(len(X_pipeline))

for bucket, regressor in regressors.items():
    mask = predicted_classes == bucket
    if mask.any():
        pipeline_predictions[mask] = regressor.predict(X_pipeline.iloc[mask])

# Простой baseline для всей validation: общая медиана train.
train_median = train_24h_df[TARGET_COLUMN].median()
baseline_predictions = np.full(len(y_pipeline), train_median)

pipeline_metrics = pd.DataFrame(
    {
        "Classifier + regressors": calculate_metrics(
            y_pipeline, pipeline_predictions
        ),
        "Overall median baseline": calculate_metrics(
            y_pipeline, baseline_predictions
        ),
    }
).T

pipeline_metrics.round(3)
```

```python
# Save models for the API without overwriting existing files.
model_dir = project_path.artifacts / "models" / "24h"
model_dir.mkdir(parents=True, exist_ok=True)
models_to_save = {"classifier": classifier_24h, **regressors}
for name, model in models_to_save.items():
    path = model_dir / f"{name}.joblib"
    if path.exists():
        print(f"Already exists: {path}")
    else:
        joblib.dump(model, path)
        print(f"Saved: {path}")
```
