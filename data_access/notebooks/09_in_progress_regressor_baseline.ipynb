## In Progress — регрессоры по диапазонам

Эмбеддинги загружаются из сохранённых выборок. Каждый регрессор оценивается на настоящем классе

```python
import pandas as pd
import numpy as np
from data_access.config import project_path
from data_access.snapshot_dataset import TARGET_COLUMN

data_dir = project_path.artifacts / "in_progress_embeddings"

train_in_progress_df = pd.read_pickle(data_dir / "train.pkl")
validation_in_progress_df = pd.read_pickle(data_dir / "validation.pkl")
test_in_progress_df = pd.read_pickle(data_dir / "test.pkl")

for name, frame in (("train", train_in_progress_df), ("validation", validation_in_progress_df), ("test", test_in_progress_df)):
    print(name, len(frame))
    
pd.DataFrame({
    "train": train_in_progress_df["Duration_Class"].value_counts(),
    "validation": validation_in_progress_df["Duration_Class"].value_counts(),
    "test": test_in_progress_df["Duration_Class"].value_counts(),
})
```

```python
train_0_1 = train_in_progress_df[train_in_progress_df["Duration_Class"] == "0-1d"]
train_1_7 = train_in_progress_df[train_in_progress_df["Duration_Class"] == "1-7d" ]
train_7_30 = train_in_progress_df[train_in_progress_df["Duration_Class"] == "7-30d"]
train_30_365 = train_in_progress_df[train_in_progress_df["Duration_Class"] == "30-365d"]

validation_0_1 = validation_in_progress_df[validation_in_progress_df["Duration_Class"] == "0-1d"]
validation_1_7 = validation_in_progress_df[validation_in_progress_df["Duration_Class"] == "1-7d" ]
validation_7_30 = validation_in_progress_df[validation_in_progress_df["Duration_Class"] == "7-30d"]
validation_30_365 = validation_in_progress_df[validation_in_progress_df["Duration_Class"] == "30-365d"]

test_0_1 = test_in_progress_df[test_in_progress_df["Duration_Class"] == "0-1d"]
test_1_7 = test_in_progress_df[test_in_progress_df["Duration_Class"] == "1-7d" ]
test_7_30 = test_in_progress_df[test_in_progress_df["Duration_Class"] == "7-30d"]
test_30_365 = test_in_progress_df[test_in_progress_df["Duration_Class"] == "30-365d"]

print("TRAIN")
print(f"  0–1 день:     {len(train_0_1)}")
print(f"  1–7 дней:     {len(train_1_7)}")
print(f"  7–30 дней:    {len(train_7_30)}")
print(f"  30–365 дней:  {len(train_30_365)}")
print(
    "  Всего по бакетам:",
    f"{sum(map(len, [train_0_1, train_1_7, train_7_30, train_30_365]))}",
    f"из {len(train_in_progress_df)}",
)

print("\nVALIDATION")
print(f"  0–1 день:     {len(validation_0_1)}")
print(f"  1–7 дней:     {len(validation_1_7)}")
print(f"  7–30 дней:    {len(validation_7_30)}")
print(f"  30–365 дней:  {len(validation_30_365)}")
print(
    "  Всего по бакетам:",
    f"{sum(map(len, [validation_0_1, validation_1_7, validation_7_30, validation_30_365]))}",
    f"из {len(validation_in_progress_df)}",
)

print("\nTEST")
print(f"  0–1 день:     {len(test_0_1)}")
print(f"  1–7 дней:     {len(test_1_7)}")
print(f"  7–30 дней:    {len(test_7_30)}")
print(f"  30–365 дней:  {len(test_30_365)}")
print(
    "  Всего по бакетам:",
    f"{sum(map(len, [test_0_1, test_1_7, test_7_30, test_30_365]))}",
    f"из {len(test_in_progress_df)}",
)
```

```python
CATEGORICAL_FEATURES = ["Project_ID", "Priority_At_Snapshot", "Type_At_Snapshot"]
NUMERICAL_FEATURES = ["Creation_Day_Of_Week", "Description_Over_100", "Story_Point_At_Snapshot", "Has_Assignee_At_Snapshot"]
EMBEDDING_FEATURES = [f"embedding_{i}" for i in range(384)]
FEATURE_COLUMNS = CATEGORICAL_FEATURES + NUMERICAL_FEATURES + EMBEDDING_FEATURES


X_train_0_1 = train_0_1[FEATURE_COLUMNS].copy()
y_train_0_1 = train_0_1[TARGET_COLUMN].copy()

X_train_1_7 = train_1_7[FEATURE_COLUMNS].copy()
y_train_1_7 = train_1_7[TARGET_COLUMN].copy()

X_train_7_30 = train_7_30[FEATURE_COLUMNS].copy()
y_train_7_30 = train_7_30[TARGET_COLUMN].copy()

X_train_30_365 = train_30_365[FEATURE_COLUMNS].copy()
y_train_30_365 = train_30_365[TARGET_COLUMN].copy()


X_validation_0_1 = validation_0_1[FEATURE_COLUMNS].copy()
y_validation_0_1 = validation_0_1[TARGET_COLUMN].copy()

X_validation_1_7 = validation_1_7[FEATURE_COLUMNS].copy()
y_validation_1_7 = validation_1_7[TARGET_COLUMN].copy()

X_validation_7_30 = validation_7_30[FEATURE_COLUMNS].copy()
y_validation_7_30 = validation_7_30[TARGET_COLUMN].copy()

X_validation_30_365 = validation_30_365[FEATURE_COLUMNS].copy()
y_validation_30_365 = validation_30_365[TARGET_COLUMN].copy()


X_test_0_1 = test_0_1[FEATURE_COLUMNS].copy()
y_test_0_1 = test_0_1[TARGET_COLUMN].copy()

X_test_1_7 = test_1_7[FEATURE_COLUMNS].copy()
y_test_1_7 = test_1_7[TARGET_COLUMN].copy()

X_test_7_30 = test_7_30[FEATURE_COLUMNS].copy()
y_test_7_30 = test_7_30[TARGET_COLUMN].copy()

X_test_30_365 = test_30_365[FEATURE_COLUMNS].copy()
y_test_30_365 = test_30_365[TARGET_COLUMN].copy()


pd.DataFrame(
    {
        "train": [
            X_train_0_1.shape,
            X_train_1_7.shape,
            X_train_7_30.shape,
            X_train_30_365.shape,
        ],
        "validation": [
            X_validation_0_1.shape,
            X_validation_1_7.shape,
            X_validation_7_30.shape,
            X_validation_30_365.shape,
        ],
        "test": [
            X_test_0_1.shape,
            X_test_1_7.shape,
            X_test_7_30.shape,
            X_test_30_365.shape,
        ],
    },
    index=["X_0_1", "X_1_7", "X_7_30", "X_30_365"],
)
```

```python
for X in (
    X_train_0_1, X_train_1_7, X_train_7_30, X_train_30_365,
    X_validation_0_1, X_validation_1_7, X_validation_7_30, X_validation_30_365,
    X_test_0_1, X_test_1_7, X_test_7_30, X_test_30_365,
):
    for column in CATEGORICAL_FEATURES:
        X[column] = X[column].fillna("Unknown").astype(str)
    for column in NUMERICAL_FEATURES:
        X[column] = pd.to_numeric(X[column], errors="raise").astype(float)
```

```python
import numpy as np


from sklearn.metrics import (
    mean_absolute_error,
    median_absolute_error,
    root_mean_squared_error,
    root_mean_squared_log_error
)


def calculate_metrics(y_true, prediction):
    prediction = np.maximum(prediction, 0.0)
    return {
        "MAE_hours": mean_absolute_error(y_true, prediction),
        "Median_AE_hours": median_absolute_error(y_true, prediction),
        "RMSE_hours": root_mean_squared_error(y_true, prediction),
        "RMSLE": root_mean_squared_log_error(y_true, prediction)
    }
```

```python
from catboost import CatBoostRegressor


regressor_params = {
    "iterations": 1000,
    "depth": 6,
    "learning_rate": 0.05,
    "loss_function": "RMSE",
    "cat_features": CATEGORICAL_FEATURES,
    "random_seed": 42,
    "allow_writing_files": False,
}

regressor_in_progress_0_1 = CatBoostRegressor(**regressor_params)
regressor_in_progress_1_7 = CatBoostRegressor(**regressor_params)
regressor_in_progress_7_30 = CatBoostRegressor(**regressor_params)
regressor_in_progress_30_365 = CatBoostRegressor(**regressor_params)
```

## Regressor in_progress - 0-1d

```python
regressor_in_progress_0_1.fit(
    X_train_0_1,
    y_train_0_1,
    eval_set=(X_validation_0_1, y_validation_0_1),
    early_stopping_rounds=100,
    use_best_model=True,
    verbose=100,
)

metrics_0_1 = {}

for split_name, features, target in (
    ("train_0_1", X_train_0_1, y_train_0_1),
    ("validation_0_1", X_validation_0_1, y_validation_0_1),
    ("test_0_1", X_test_0_1, y_test_0_1),
):
    prediction = regressor_in_progress_0_1.predict(features)
    metrics_0_1[split_name] = calculate_metrics(target, prediction)
```

```python
median_0_1 = y_train_0_1.median()
baseline_prediction_0_1 = np.full(
    len(y_test_0_1),
    median_0_1,
)

baseline_metrics_0_1 = calculate_metrics(
    y_test_0_1,
    baseline_prediction_0_1,
)
```

## Every regressor is in its own current bucket

```python
direct_metrics = pd.concat(
    {
        "0-1d": pd.DataFrame(metrics_0_1).T,
        "1-7d": pd.DataFrame(metrics_1_7).T,
        "7-30d": pd.DataFrame(metrics_7_30).T,
        "30-365d": pd.DataFrame(metrics_30_365).T,
    },
    names=["Bucket", "Split"],
).round(3)

direct_metrics
```

```python
median_metrics = pd.DataFrame.from_dict(
    {
        "0-1d": baseline_metrics_0_1,
        "1-7d": baseline_metrics_1_7,
        "7-30d": baseline_metrics_7_30,
        "30-365d": baseline_metrics_30_365,
    },
    orient="index",
)

median_metrics.index.name = "Bucket"

direct_test = direct_metrics[
    direct_metrics.index.get_level_values("Split").str.startswith("test_")
].droplevel("Split")

comparison = pd.concat(
    {"CatBoost": direct_test, "Median baseline": median_metrics},
    axis=1,
)

comparison.round(3)
```
